This Jupyter Notebook sets up a workspace directory and mounts Google Drive when running on Colab.


## Initialize Workspace

We'll check if we're running in Google Colab, and if so, we'll mount Google Drive and keep the workspace there. Otherwise, the workspace is the local `/workspace` directory.


```python
import sys
from pathlib import Path

if "google.colab" in sys.modules:
    from google.colab import drive

    drive.mount("/content/drive")
    workspace_dir = Path("/content/drive/MyDrive/workspace")
else:
    workspace_dir = Path("/workspace")

project_dir = workspace_dir / "projects/aibasics/examples"
print("project_dir:", project_dir)
```
